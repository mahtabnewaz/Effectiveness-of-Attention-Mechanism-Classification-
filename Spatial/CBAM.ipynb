```python
# CBAM Block
class CBAM(nn.Module):
    def __init__(self, in_channels, kernel_size=7):
        super(CBAM, self).__init__()

        # Spatial Attention
        self.spatial_attention = nn.Sequential(
            nn.Conv2d(2, 1, kernel_size=kernel_size, stride=1, padding=kernel_size // 2, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x):
        # Apply spatial attention
        max_pool = torch.max(x, dim=1, keepdim=True)[0]  # Max pooling along channel dimension
        avg_pool = torch.mean(x, dim=1, keepdim=True)   # Average pooling along channel dimension
        pooled = torch.cat([max_pool, avg_pool], dim=1) # Concatenate along channel dimension

        spatial_attention = self.spatial_attention(pooled)
        return x * spatial_attention
```
